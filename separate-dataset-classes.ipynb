# Dataset Training

```python
import pandas as pd
import os
import shutil
import sys

# === KONFIGURASI ===
csv_path = '/home/ahmadwafii11/Documents/Skripsi/Tugas-Akhir-Hama-Penyakit-Kakao.v45i.tensorflow/train/_annotations.csv'
image_dir = '/home/ahmadwafii11/Documents/Skripsi/Tugas-Akhir-Hama-Penyakit-Kakao.v45i.tensorflow/train'
output_dir = '/home/ahmadwafii11/Documents/Skripsi/dataset/sekunder1'

# === BACA CSV ===
try:
    df = pd.read_csv(csv_path)
except Exception as e:
    print(f'Gagal membaca file CSV: {e}')
    sys.exit(1)

# Ambil kolom yang dibutuhkan
if 'filename' not in df.columns or 'class' not in df.columns:
    print("CSV tidak mengandung kolom 'filename' dan 'class'")
    sys.exit(1)

df = df[['filename', 'class']]
df['class'] = df['class'].str.strip()  # Hilangkan spasi berlebih

# === PROSES FILE ===
for idx, row in df.iterrows():
    filename = row['filename']
    label = row['class']

    src_path = os.path.join(image_dir, filename)
    class_folder = os.path.join(output_dir, label)
    dst_path = os.path.join(class_folder, filename)

    # Jika file tidak ditemukan → STOP
    if not os.path.exists(src_path):
        print(f"❌ File tidak ditemukan: {src_path}")
        sys.exit(1)

    # Buat folder jika belum ada
    os.makedirs(class_folder, exist_ok=True)

    # Salin file
    shutil.copy2(src_path, dst_path)

print("✅ Semua file berhasil dipindahkan ke folder sesuai kelas.")
```
