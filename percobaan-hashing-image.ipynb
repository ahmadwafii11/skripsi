```python
from pathlib import Path
from PIL import Image
import imagehash
import os
```

```python
def remove_duplicate_images(
    dataset_dir,
    hash_size=16,
    extensions=(".jpg", ".jpeg"),
):
    """
    Mencari dan menghapus gambar duplikat berdasarkan perceptual hash.

    Parameters
    ----------
    dataset_dir : str
        Folder dataset.
    hash_size : int
        Ukuran hash (default=16).
    extensions : tuple
        Format gambar yang akan diproses.
    """

    dataset_dir = Path(dataset_dir)

    hashes = {}
    deleted_count = 0

    for image_path in dataset_dir.rglob("*"):

        if image_path.suffix.lower() not in extensions:
            continue

        try:
            with Image.open(image_path) as img:
                img_hash = str(imagehash.phash(img, hash_size=hash_size))

            if img_hash in hashes:

                original_file = hashes[img_hash]

                print(
                    f"[DUPLIKAT] {image_path.name}"
                    f" -> sama dengan {original_file.name}"
                )

                os.remove(image_path)
                deleted_count += 1

            else:
                hashes[img_hash] = image_path

        except Exception as e:
            print(f"Error {image_path}: {e}")

    print("\n===== SELESAI =====")
    print(f"Jumlah gambar unik : {len(hashes)}")
    print(f"Jumlah duplikat dihapus : {deleted_count}")
```

```python
remove_duplicate_images("dataset")
```
